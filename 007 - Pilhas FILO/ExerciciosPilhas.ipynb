# Projeto 1 – Histórico de Filtros em Dashboard de Dados 
Contexto: Em aplicações de ciência de dados para negócios, usuários interagem com dashboards aplicando filtros e transformações aos dados (por exemplo, em tabelas ou gráficos interativos). É útil permitir que o usuário desfaça filtros aplicados (por exemplo, revertendo ao estado anterior dos dados) sem recarregar tudo do início. 

Problema: Como implementar um mecanismo de histórico de operações que guarde estados anteriores dos dados ao aplicar filtros, para que o usuário possa desfazer a última ação? 

```python
class Dashboard:
    def __init__(self):
        self.filters = []

    def add_filter(self, newfilter):
        self.filters.append(newfilter)

    def filters_list(self):
        for item in self.filters:
            print(item)

    def undo(self):
        self.filters.pop(0)
```

```python
HRDashboard = Dashboard()

HRDashboard.add_filter("Ano")
```

```python
HRDashboard.filters_list()
```

```python
HRDashboard.add_filter("BU")

HRDashboard.add_filter("Produto")

HRDashboard.add_filter("Categoria")
```

```python
HRDashboard.filters_list()
```

```python
HRDashboard.undo()
HRDashboard.filters_list()
```

# Projeto 2 – Calculadora de Expressões Aritméticas 
Contexto: Muitas calculadoras e programas de matemática interpretam expressões aritméticas infixas (como "3 + 4 * (2 - 1)"). Uma maneira clássica de avaliar tais expressões é converter para notação pós-fixa(RPN) ou usar pilhas para operar em tempo de execução. 

Problema: Escrever um programa Python que avalie uma expressão matemática em infixo contendo +, -, *, /, e parênteses. Queremos usar pilhas para lidar com a precedência de operadores e a aninhamento deparênteses. 

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if len(self.items) > 0:
            return self.items.pop()

    def is_empty(self):
        return len(self.items) == 0

    def peek(self):
        if len(self.items) > 0:
            return self.items[-1]

def calcular(expressao):
    numeros = Stack()
    operadores = Stack()

    # PRIORIZAÇÃO
    prioridade = {"+": 1, "-": 1, "*": 2, "/": 2}


    def fazer_conta():
        operador = operadores.pop()

        numero2 = numeros.pop()
        numero1 = numeros.pop()

        if operador == "+":
            resultado = numero1 + numero2
        elif operador == "-":
            resultado = numero1 - numero2
        elif operador == "*":
            resultado = numero1 * numero2
        elif operador == "/":
            resultado = numero1 / numero2

        numeros.push(resultado)

    i = 0


    while i < len(expressao):
        caractere = expressao[i]

        if caractere == " ":
            i += 1
            continue

        if caractere.isdigit():
            numero = ""

            while i < len(expressao) and expressao[i].isdigit():
                numero += expressao[i]
                i += 1

            numeros.push(float(numero))
            continue


        if caractere == "(":
            operadores.push(caractere)

        elif caractere == ")":
            while operadores.peek() != "(":
                fazer_conta()
            operadores.pop()

        elif caractere in "+-*/":
            while (
                not operadores.is_empty()
                and operadores.peek() != "("
                and prioridade[operadores.peek()] >= prioridade[caractere]
            ):

                fazer_conta()
            operadores.push(caractere)
        i += 1


    while not operadores.is_empty():
        fazer_conta()
    return numeros.pop()


```

```python
expressao = "3 + 4 * (2 - 1)"
resultado = calcular(expressao)

print(f"Resultado: {resultado}")
```
