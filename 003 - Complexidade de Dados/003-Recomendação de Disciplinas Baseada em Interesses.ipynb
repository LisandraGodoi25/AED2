# Func Aux

```python
def addAlunos():
    addName = input("Insira o nome do novo aluno: ")

    materiasss = []
    
    while True:

        materia = input("Insira o nome de uma matéria cursada ou QUIT para parar:").upper()
        if materia == "QUIT":
            break
        materiasss.append(materia)

    alunos[addName] = materiasss
    lista_alunos.append(addName)

    print(f"Aluno adicionado com sucesso: {addName}: {alunos[addName]}")
```

```python
def sugerirMaterias():

    ALUNO = input("Qual aluno deseja sugestões de matérias? ")
    MATERIASJACURSADAS = alunos[ALUNO]
    materiasFILTRADAS = []

    for subject in MATERIASJACURSADAS:
        for student in lista_alunos:

            if subject in alunos[student]:
                materiasFILTRADAS = materiasFILTRADAS + alunos[student]

    materiasFILTRADASSEMDUPLICATAS = list(set(materiasFILTRADAS))
    
    for subject in MATERIASJACURSADAS:
        materiasFILTRADASSEMDUPLICATAS.remove(subject)

    print(f"{ALUNO} já cursou:")
    print(MATERIASJACURSADAS)
    print("As sugestões de matérias são:")
    print(materiasFILTRADASSEMDUPLICATAS)
```
