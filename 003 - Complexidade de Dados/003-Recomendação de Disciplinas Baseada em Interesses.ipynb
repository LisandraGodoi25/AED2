# Projeto 2: Recomendação de Disciplinas Baseada em Interesses 
Contexto: A Fatec Votorantim oferece diversas disciplinas optativas. Para ajudar os alunos a escolherem, a coordenação quer criar um sistema simples de recomendação. Dado o histórico de disciplinas que um aluno já cursou, o sistema deve sugerir outras disciplinas que alunos com interesses semelhantes também cursaram. 

Problema: Dada uma lista de históricos de disciplinas de alunos (onde cada histórico é uma lista de códigos de disciplinas) e o histórico de um aluno específico, implemente uma função em Python para encontrar disciplinas que outros alunos com pelo menos uma disciplina em comum cursaram, mas que o aluno original ainda não 
cursou.

Indicação de como resolver: Para cada outro aluno, compare seu histórico de disciplinas com o do aluno 
original. Se houver alguma disciplina em comum, adicione as disciplinas que o outro aluno cursou e que o aluno original não cursou a uma lista de recomendações. Analise a complexidade de tempo dessa abordagem.
