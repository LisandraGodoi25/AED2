## **Exercício 2:**
Analise a complexidade da função encontrar_maximo(lista) que retorna o maior valor em uma lista: 
```
def encontrar_maximo(lista): 
    valor_maximo = lista[0] #1
    for numero in lista: #n
        if numero > valor_maximo: #1
            valor_maximo = numero #1
    return valor_maximo 
```
A lista de entrada tem n elementos. Qual a ordem de complexidade em função de n?

## R. A complexidade é O(n)

```
def busca_binaria(lista, alvo): 
    esquerda = 0 
    direita = len(lista) - 1 
        while esquerda <= direita: 
            meio = (esquerda + direita) // 2 
            if lista[meio] == alvo: 
                return meio 
            elif lista[meio] < alvo: 
                esquerda = meio + 1 
            else: 
                direita = meio - 1 
return -1 
```

## Exercício 5: 
A função abaixo calcula o n-ésimo número de Fibonacci de forma recursiva (definindo fib(0)=0, fib(1)=1): 
```
def fib(n): 
    if n < 2: 
        return n 
    else: 
        return fib(n-1) + fib(n-2) 
```
Qual é a complexidade de tempo dessa implementação em termos de n?

### R. O(n^2)
