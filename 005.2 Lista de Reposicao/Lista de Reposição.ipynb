# Exercício 5 — Conta simples com encapsulamento  [Fácil] 
Crie uma classe Conta que mantenha o saldo protegido contra alteração direta. 

Requisitos: 
- O construtor recebe o nome do titular e inicia o saldo em 0. 
- Use __saldo como atributo privado. 
- Implemente depositar(valor), permitindo apenas valores positivos. 
- Implemente consultar_saldo(), retornando o saldo atual. 
- Crie uma conta, faça dois depósitos e mostre o saldo

```python
from rich import print
from rich.panel import Panel
from abc import ABC, abstractclassmethod

class Conta():
    def __init__(self, titular):
        self.titular = titular
        self.__saldo = 0

    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
        
        print(f"R${valor} depositado com sucesso")

    def consultar_saldo(self):
        print(f"Saldo de {self.titular} é de R${self.__saldo}")
```

```python
meubanquinho = Conta("Lys")

meubanquinho.depositar(100)
meubanquinho.depositar(140)
meubanquinho.consultar_saldo()
```

# Exercício 6 — Animais: herança e polimorfismo  [Fácil] 
Implemente uma pequena hierarquia de animais para exercitar herança e polimorfismo. 
Algoritmos e Estrutura de Dados II  •  Aula 05 — Programação Orientada a Objetos 
Requisitos: 

- Crie a classe Animal com atributo nome e método emitir_som(). 
- Crie Cachorro e Gato herdando de Animal. 
- Redefina emitir_som(): Cachorro retorna “au au” e Gato retorna “miau”. 
- Crie uma lista com um cachorro e um gato e percorra-a chamando emitir_som() sem testar o tipo do 
objeto. 

```python
from rich import print
from rich.panel import Panel
from abc import ABC, abstractclassmethod

class Animal(ABC):
    def __init__(self, nome):
        self.nome = nome

    @abstractclassmethod
    def emitir_som(self):
        pass



class Cachorro(Animal):
    def __init__(self, nome):
        super().__init__(nome)

    def emitir_som(self):
        return f"{self.nome}: au au AU"



class Gato(Animal):
    def __init__(self, nome):
        super().__init__(nome)

    def emitir_som(self):
        return f"{self.nome}: Miauuuuuuuu"
```

```python
animallist = [Cachorro("Meg"),Gato("Mocha")]

for pets in animallist:
    print(pets.emitir_som())
```

# Exercício 7 — Veículo com velocidade controlada  [Média] 
Modele um veículo inspirado no exemplo Automovel da aula, mas com regras completas de aceleração e frenagem. 
Requisitos: 
- A classe Veiculo deve ter placa, velocidade_max e velocidade_atual; torne os três atributos privados. 
- velocidade_atual começa em 0. 
- Crie getters para placa e velocidade_max. 
- Crie um setter para velocidade_max que aceite apenas valores positivos e nunca deixe a nova máxima 
menor que a velocidade atual. 
- acelerar(incremento=10) aumenta a velocidade sem ultrapassar a máxima. 
- frear(decremento=10) reduz a velocidade sem ficar negativa. 
- Implemente __str__ para mostrar placa, velocidade atual e máxima. 
- Demonstre o funcionamento com várias acelerações e frenagens.

```python
from rich import print
from rich.panel import Panel
from abc import ABC, abstractclassmethod

class Veiculo():
    def __init__(self, placa, velocidade_max):
        self.__placa = placa
        self.__velMax = velocidade_max
        self.__velAtual = 0

    @property
    def placa (self):
        return self.__placa

    @property
    def velocidade_atual(self):
        return self.__velAtual

    @property
    def velocidade_max(self):
        return self.__velMax
    
    @velocidade_max.setter
    def velocidade_max(self, new):
        if new > 0 and new > self.__velAtual:
            self.__velMax = new
            print(f"Velocidade máxima definida para {self.__velMax}")
        else:
            print("Velocidade máxima sugerida não cumpre as regras estabelecidas")

    def acelerar(self, incremento = 10):
        if (self.__velAtual + incremento) < self.__velMax:
            self.__velAtual += incremento
        else:
            self.__velAtual = self.__velMax

    def frear(self, decremento = 10):
        if (self.__velAtual - decremento) > 0:
            self.__velAtual -= decremento
        else:
            self.__velAtual = 0

    def __str__(self):
        conteudo = f"Placa: {self.__placa}"
        conteudo += f"\nVelocidade Atual: {self.__velAtual}"
        conteudo += f"\nVelocidade Máxima: [black on red]{self.__velMax}[/]"

        info = Panel(conteudo, title="Informações do Carro", width=30)

        return info
```

```python
carrin = Veiculo("123ABC", 80)

carrin.acelerar(70)

carrin.acelerar(70)

carrin.velocidade_atual

carrin.frear(100)

carrin.acelerar(14)

carrin.acelerar(20)

carrin.frear(6)

carrin.velocidade_atual
```

# Exercício 8 — Funcionários e cálculo de bônus  [Média] 
Crie uma hierarquia para calcular bônus de funcionários de uma empresa. 
Requisitos: 
- Funcionario possui nome e salario; implemente __str__. 
- Funcionario.calcular_bonus() retorna 5% do salário. 
- Vendedor herda de Funcionario, acrescenta vendas e redefine calcular_bonus() para 5% do salário + 2% 
das vendas. 
- Gerente herda de Funcionario, acrescenta equipe e redefine calcular_bonus() para 10% do salário + R$ 
100 por integrante da equipe. 
- Crie pelo menos um objeto de cada tipo, coloque-os em uma única lista e exiba nome, tipo e bônus 
usando o mesmo método calcular_bonus(). 

```python
from rich import print
from rich.panel import Panel
from abc import ABC, abstractclassmethod

class Funcionario(ABC):
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario
        self.bonus = 0
        self.tipo = "Funcionário"

    def __str__(self):
        return f"Nome: {self.nome}\nSalário: R${self.salario:,.2f}"

    def calcular_bonus(self):
        self.bonus = self.salario / 100 * 5

class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas
        self.tipo = "Vendedor"

    def calcular_bonus(self):
        self.bonus = (self.salario / 100 * 5) + (self.vendas / 100 * 2)

class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe
        self.tipo = "Gerente"

    def calcular_bonus(self):
        self.bonus = (self.salario / 100 * 10) + (100 * self.equipe)
```

```python
employees = [Funcionario("Lisandra Godoi", 2000), Vendedor("Maria Leopoldina", 3500, 24530), Gerente("Bruna Ferreira", 12340, 6)]

for employee in employees:
    employee.calcular_bonus()
    conteudo = f"Nome: {employee.nome}"
    conteudo += f"\nTipo: {employee.tipo}"
    conteudo += f"\nBônus: R${employee.bonus:,.2f}"

    ficha = Panel(conteudo, title="Ficha Institucional", width=30)

    print(ficha)
```

# Exercício 9 — Produtos e contador de objetos  [Média] 
Crie uma classe Produto para um catálogo de e-commerce, explorando atributo de classe e encapsulamento.  Requisitos: 
- Crie o atributo de classe contador iniciado em 0. 
- Cada produto deve receber automaticamente um id sequencial a partir do contador. 
- Use atributos privados para id, nome e preco. 
- Crie getters para id, nome e preco. 
- Crie set_preco(novo_preco), aceitando apenas valores maiores que zero. 
- Implemente aplicar_desconto(percentual), validando percentual entre 0 e 100. 
- Implemente __str__. 
- Crie três produtos e, ao final, mostre quantos objetos foram criados usando Produto.contador

```python
from rich import print
from rich.panel import Panel
from abc import ABC, abstractclassmethod

class Produto(ABC):
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    @property
    def showid(self):
        return self.__id

    @property
    def shownome(self):
        return self.__nome
    
    @property
    def showpreco(self):
        return self.__preco

    def set_preco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("O preço informado precisa ser maior que zero")

    def aplicar_desconto(self, percentual):
        if percentual >= 0 and percentual <= 100:
            desconto = self.__preco / 100 * percentual
            print(f"Desconto de [black on white]{percentual}%[/] aplicado")
            self.__preco -= desconto
            print(f"Novo preço: R${self.__preco:,.2f}")
        else: 
            print("Insira um desconto entre 1% e 100%")

    def __str__(self):
        print(f"Produto: {self.__nome}")
        print(f"Preço: R${self.__preco:,.2f}")
```

# Exercício 10 — Mini sistema de pedidos de e-commerce  [Difícil] 
Desenvolva um pequeno sistema orientado a objetos para pedidos de um e-commerce. O objetivo é integrar a 
maior parte dos conceitos trabalhados na aula em um único problema. 
Requisitos: 
- Crie uma classe Produto com id automático (atributo de classe), nome e preço privados; inclua getters, setter validado para preço e __str__. 
- Crie ProdutoFisico herdando de Produto, com peso, e ProdutoDigital herdando de Produto, com tamanho_mb. 
- Em Produto, crie calcular_frete() retornando 0. Redefina-o em ProdutoFisico para retornar R$ 5,00 + R$ 
2,00 por kg. ProdutoDigital deve manter frete igual a 0. Essa chamada comum deve demonstrar polimorfismo. 
- Crie uma classe Cliente com nome e email privados, getters e __str__. 
- Crie uma classe Pedido com número automático (atributo de classe), cliente, lista de produtos e status privado. 
- Pedido deve oferecer adicionar_produto(produto), calcular_total(), calcular_frete_total(),fechar_pedido() e __str__. 
- Depois de fechado, o pedido não deve aceitar novos produtos. Essa regra deve ser controlada internamente pela classe. 
- calcular_total() deve somar os preços dos produtos. calcular_frete_total() deve chamar calcular_frete() de cada item sem testar se ele é físico ou digital. 
- No programa principal, crie um cliente, ao menos três produtos (misturando físicos e digitais), monte um pedido, exiba o resumo, feche-o e tente adicionar mais um produto para comprovar a regra.

```python
from rich import print
from rich.panel import Panel
from abc import ABC, abstractclassmethod

class Produto(ABC):
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    @property
    def showid(self):
        return self.__id

    @property
    def shownome(self):
        return self.__nome
    
    @property
    def showpreco(self):
        return self.__preco

    @showpreco.setter
    def showpreco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("O preço informado precisa ser maior que zero")

    def __str__(self):
        print(f"Produto: {self.__nome}")
        print(f"Preço: R${self.__preco:,.2f}")

    def frete(self):
        return 0


class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.peso = peso

    def frete(self):
        return 5 + (2 * self.peso)


class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb


class Cliente:
    def __init__(self, nome, email):
        self.__nome = nome
        self.__email = email

    @property
    def nome(self):
        return self.__nome

    @property
    def email(self):
        return self.__email
    
    def __str__(self):
        print(f"Produto: {self.__nome}")
        print(f"Email: R${self.__email}")
    

class Pedido():
    contador = 0

    def __init__(self, cliente):
        Pedido.contador += 1
        self.__id = Pedido.contador
        self.cliente = cliente
        self.listadeprodutos = []
        self.__status = "Aberto"


    def adicionar_produto(self, newproduto):
        if self.__status == "Aberto":
            self.listadeprodutos.append(newproduto)
        else:
            print("Não é possível adicionar itens em pedidos finalizados")

    
    def calcular_total(self):
        precototal = 0

        for item in self.listadeprodutos:
            #print(f"Item atual: {item.shownome}")
            precototal += item.showpreco

        return f"Preço total: R${precototal:,.2f}"


    def calcular_frete_total(self):
        fretetotal = 0

        for item in self.listadeprodutos:
            frete = item.frete()
            #print(f"Frete da {item.shownome}: {frete}")
            fretetotal += frete
            self.__freteTotal = fretetotal
        return f"Frete Total: R${fretetotal:,.2f}"

    def fechar_pedido(self):
        self.__status = "Fechado"

    def __str__(self):
        conteudo = f"Pedido: {self.__id}"
        conteudo += f"\nStatus: {self.__status}"
        conteudo += f"\nItens:"
        for item in self.listadeprodutos:
            conteudo += f"\n     :blue_circle: {item.shownome}"
        conteudo += f"\n{self.calcular_total()}"
        conteudo += f"\n{self.calcular_frete_total()}"

        fichastr = Panel(conteudo, title=f"Pedido {self.__id}", width=50)
        print(fichastr)
```

```python
pedidin = Pedido("Lisandra Godoi")
```
