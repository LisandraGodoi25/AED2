## Exercício 5 - Calculando a altura da árvore (Prática)

**Enunciado:**
Escreva uma função `altura(raiz)` em Python que recebe a raiz de uma árvore binária e retorna a altura da árvore (número de níveis). Dica: use recursão. Em seguida, teste-a na árvore criada no exemplo do capítulo.

````python
class No:
    def __init__(self, valor):
        self.valor = valor
        self.esquerda = None
        self.direita = None

# Árvore do exemplo do capítulo
raiz = No('a')
raiz.esquerda = No('b')
raiz.direita = No('c')
raiz.esquerda.esquerda = No('d')
raiz.esquerda.direita = No('e')

def altura(no):
    if no is None:
        return 0
    alt_e = altura(no.esquerda)
    alt_d = altura(no.direita)
    return 1 + max(alt_e, alt_d)

print("Altura da árvore (número de níveis):", altura(raiz))
````

## Exercício 6 - Contando nós folha (Prática)

**Enunciado:**
Implemente em Python uma função `conta_folhas(raiz)` que receba a raiz de uma árvore binária e retorne o número total de nós folha. Mostre a chamada da função no exemplo do capítulo.

````python
def conta_folhas(no):
    if no is None:
        return 0
    if no.esquerda is None and no.direita is None:
        return 1
    return conta_folhas(no.esquerda) + conta_folhas(no.direita)

print("Quantidade de nós folha:", conta_folhas(raiz))
````

## Projeto 1 - Árvore de Decisão para Negócios

**Enunciado:**
Dado um conjunto de clientes rotulados como bom (`1`) ou ruim (`0`) pagador com base em `[idade, renda]`, construir um modelo de árvore de decisão (`DecisionTreeClassifier`) que preveja o risco de um novo cliente.

````python
from sklearn import tree

# Cada item de X é [idade, renda]
# Em y: 0 = ruim pagador e 1 = bom pagador
X = [
    [25, 5000],
    [40, 12000],
    [22, 3000],
    [35, 8000],
    [50, 10000]
]
y = [1, 1, 0, 1, 0]

clf = tree.DecisionTreeClassifier()
clf = clf.fit(X, y)

# Novo cliente: 30 anos e renda 7000
previsao = clf.predict([[30, 7000]])
print("Previsão para novo cliente [30, 7000]:", previsao)
````

## Projeto 2 - Árvore Genealógica Interativa

**Enunciado:**
Criar uma estrutura em Python que represente membros de uma família e permita exibir a árvore de parentesco hierarquicamente a partir de um ancestral.

````python
familia = {
    'João': ['Pedro', 'Ana'],
    'Pedro': ['Carlos', 'Marina'],
    'Ana': ['Julia'],
    'Carlos': [],
    'Marina': [],
    'Julia': []
}

def exibir_arvore(nome, nivel=0):
    print('  ' * nivel + nome)
    for filho in familia.get(nome, []):
        exibir_arvore(filho, nivel + 1)

exibir_arvore('João')
````
