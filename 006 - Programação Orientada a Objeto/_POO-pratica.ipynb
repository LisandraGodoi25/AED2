```python
# Melhorando a classe Automóvel
class Automovel:
    def __init__(self, placa, velocidade_maxima):
        self.placa = placa
        self.velocidade_maxima = velocidade_maxima

    #def to_str(self):
     #   return f'{self.velocidade_maxima} km/h'

    def get_placa(self): # GET PEGA PROPRIEDADES INTERNAS
        return self.placa

    #metodo mágico __tostr__
    def __tostr__():
        return f'{self.velocidade_maxima} km/h'
    
    def dirigir(self, velocidade):
        print(f"Estou digirindo a velocidade de {velocidade} km/h")
```

```python
meucarro = Automovel(1234, 300)

print(f"Meu carro de placa {meucarro.placa}")
```

```python
meucarro.dirigir(200)
```
