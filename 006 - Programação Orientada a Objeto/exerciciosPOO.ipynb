5.  (PRÁTICA):  Classe  Pessoa:  Crie  uma  classe  Pessoa  com  nome  e  idade.  Adicione  método 
faz_aniversario() que incrementa a idade em 1. Use __str__() para retornar algo como "<nome>, <idade> 
anos". Crie duas pessoas e mostre a idade de cada uma antes e depois de chamar faz_aniversario()

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def fazer_aniversario(self):
        self.idade += 1

    def __str__(self):
        print(f"{self.nome}, {self.idade} anos")
````

````python
pessoa1 = Pessoa("Ava", 19)
pessoa2 = Pessoa("Beatrice", 23)

pessoa1.__str__()
pessoa1.fazer_aniversario()
pessoa1.__str__()


pessoa2.__str__()
pessoa2.fazer_aniversario()
pessoa2.__str__()
````

7. (PRÁTICA): Herança: Implemente duas classes com herança: 
• Classe base Animal com um método fazer_som() que imprime algo genérico, por exemplo 
"Algum som". 
• Duas subclasses Cachorro e Gato que herdam de Animal. Na classe Cachorro, sobrescreva 
fazer_som()  para  imprimir  "Au  Au".  Em  Gato,  para  "Miau". 
Crie um objeto de cada e chame fazer_som() em cada um. 

````python
from abc import ABC, abstractclassmethod

class Animal(ABC):
    def __init__(self):
        pass

    @abstractclassmethod
    def fazer_som():
        pass


class Cachorro(Animal):
    def __init__(self):
        super().__init__()

    def fazer_som(self):
        print("Au auuuuuuu")

class Gato(Animal):
    def __init__(self):
        super().__init__()

    def fazer_som(self):
        print("Miauuuuuuuuu")
````

````python
gatin = Gato()

doginho = Cachorro()
````

````python
gatin.fazer_som()
````

````python
doginho.fazer_som()
````

8.  (PRÁTICA):  Polimorfismo:  Dada  as  classes  do  Exercício  4,  escreva  uma  função  externa 
animais_falam(lista_animais) que recebe uma lista de objetos (que podem ser Cachorro ou Gato) e 
chama fazer_som() em cada um. Mostre que a função pode operar sobre diferentes tipos de animais 
(polimorfismo em ação). 

````python
def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()

animais = [Cachorro(), Gato(), Cachorro()]

animais_falam(animais)
````
