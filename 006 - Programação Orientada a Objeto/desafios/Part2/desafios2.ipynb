```python
from abc import ABC, abstractclassmethod

class poligono(ABC):

    def __init__(self):
        self.lados = 0

    @abstractclassmethod
    def perimetro(self):
        pass

    def area(self):
        pass


class Quadrado(poligono):
    def __init__(self, width):
            self.width = width

    def perimetro(self):
        print(f"Perimetro: {self.width * 4}")
    def area(self):
        print(f"Area: {self.width * self.width}")
    
class Circulo(poligono):
    def __init__(self, raio):
        self.raio = raio
        
    def perimetro(self):
        print(f"Perimetro: {2 * 3.14 * self.raio}")

    def area(self):
        print(f"Area: {3.14 * (self.raio * self.raio)}")
            

```

```python
circulinho = Circulo(2)
circulinho.perimetro()

quadradinho = Quadrado(8)
quadradinho.area()
```

# Cafeteira que faz cafe, cha e leite quente
metodos: 
preparar, ferver agua
abstratos: misturar, servir



```python
from abc import ABC, abstractclassmethod
from time import sleep
from rich import print

class BebidaQuente(ABC):
    def __init__():
        pass

    def preparar(self):
        self.ferverAgua()
        self.misturar()
        sleep(2)
        self.servir()
        sleep(2)
        print("Sua bebida está pronta! Obrigada pela preferência! :green_circle:")

    def ferverAgua(self):
        print("Fervendo a agua 1/10")
        sleep(2)
        print("Fervendo a agua 6/10")
        sleep(2)
        print("Fervendo a agua 10/10")
        sleep(2)
        print("Agua fervida com sucesso! :green_circle:")
        sleep(2)

    @abstractclassmethod
    def misturar(self):
        pass

    @abstractclassmethod
    def servir(self):
        pass
```

```python
class Cafe(BebidaQuente):
    def __init__(self, qtd):
        self.qtd = qtd
    
    def misturar(self):
        print("Colocando o café... ")
        sleep(2)
        print("Colocando a água...")
        sleep(2)
        print("misturando")

    def servir(self):
        print(f"Servindo {self.qtd} café(s) na chícara com açúcar :coffee:")

class Cha(BebidaQuente):
    def __init__(self, qtd):
        self.qtd = qtd
    
    def misturar(self):
        print("Colocando as folhas...")
        sleep(2)
        print("Colocando o açúcar...")
        sleep(2)
        print("Colocando a água...")
        sleep(2)
        print("misturando")

    def servir(self):
        print(f"Servindo {self.qtd} chá(s) na chícara com mel :tea:")


class Leite(BebidaQuente):
    def __init__(self, qtd):
        self.qtd = qtd
    
    def misturar(self):
        print("Colocando o leite...")
        sleep(2)
        print("misturando")

    def servir(self):
        print(f"Servindo {self.qtd} leite(s) na caneca :milk:")
```

```python
cafezin = Cafe(2)

cafezin.preparar()
```

```python
chazin = Cha(1)

chazin.preparar()
```

```python
leitin = Leite(1)

leitin.preparar()
```

# Calcular fretes de veiculos diferentes:
Transporte - abc
distancia
frete

calFrete() - abc

Moto fator 0.5
Caminhao 1.20
Drone 9.5

```python
from abc import ABC, abstractclassmethod
from time import sleep
from rich import print

class Transporte(ABC):
    def __init__(self, distancia):
        self.distancia = distancia
        self.frete = 0

    @abstractclassmethod
    def calFrete(self):
        pass


class Moto(Transporte):
    def __init__(self, distancia, fator = 0.5):
        self.distancia = distancia
        self.fator = fator

    def calFrete(self):
        self.frete = self.fator * self.distancia
        print(f"FRETE MOTO: {self.frete}")

class Caminhao(Transporte):
    def __init__(self, distancia, fator = 2.5):
        self.distancia = distancia
        self.fator = fator

    def calFrete(self):
        if self.distancia < 50:
            print("Distancia mínima de 50 km não alcançada :red_circle:")
        else:
            self.frete = self.fator * self.distancia
            print(f"FRETE ONIBUS: {self.frete}")


class Drone(Transporte):
    def __init__(self, distancia, fator = 9.5):
        self.distancia = distancia
        self.fator = fator

    def calFrete(self):
        if self.distancia > 10:
            print("Distancia máxima de 10km ULTRAPASSADA :red_circle:")
        else:
            self.frete = self.fator * self.distancia
            print(f"FRETE DRONE: {self.frete}")
```

```python
motin = Moto(20)

motin.calFrete()

dronin = Drone(15)
dronin.calFrete()
```

```python
from abc import ABC, abstractclassmethod
from time import sleep
from rich import print

class Funcionario(ABC):
    def __init__(self, nome, salMinimo = 1612, inss = 7.5):
        self.nome = nome
        self.salBruto = 0
        self.salLiquido = 0
        self.salMinimo = salMinimo
        self.inss = inss
    
    @abstractclassmethod
    def calcularSalario(self):
        pass

    def analisarSalario(self):
        aux = self.salLiquido / self.salMinimo

        print(f"Seu salário é {aux:,.2f} o salário mínimo")

class Horista(Funcionario):
    def __init__(self, nome, qtdHoras, precoHora):
        super().__init__(nome)
        self.qtdHoras = qtdHoras
        self.precoHora = precoHora

    def calcularSalario(self):
        self.salBruto = self.qtdHoras * self.precoHora
        self.salLiquido = self.salBruto - (self.salBruto / 100 * self.inss)
        print(f"O seu salário recebido será de [black on white]R${self.salLiquido:,.2f}[/]")

class Mensalista(Funcionario):
    def __init__(self, nome, salBruto):
        super().__init__(nome)
        self.salBruto = salBruto
    
    def calcularSalario(self):
        self.salLiquido = self.salBruto - (self.salBruto / 100 * self.inss)
        print(f"O seu salário recebido será de [black on white]R${self.salLiquido:,.2f}[/]")

```

```python
horinha = Horista("Bruno", 10,210)
horinha.calcularSalario()
horinha.analisarSalario()
```

```python
mensalzinho = Mensalista("Ana", 3000)

mensalzinho.calcularSalario()
mensalzinho.analisarSalario()
```
