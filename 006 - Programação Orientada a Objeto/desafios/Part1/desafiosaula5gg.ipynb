# Criar funcionario
Com nome, setor e cargo, e metodo para se apresentar

```python
class Employee:

    def __init__(self, nome, cargo, setor):
        self.nome = nome
        self.cargo = cargo
        self.setor = setor
        print(f"Novo funcionário criado com sucesso!")

    def apresentacao(self):
        return f"Olá, meu nome é {self.nome}, trabalo no setor de {self.setor} e sou {self.cargo}"

```

```python
lisandra = Employee("Lisandra", "Assistente", "Qualidade")

print(lisandra.apresentacao())
```

# Crie um produto
Nome e Preço e Método que mostra a etiqueta do produto

```python
class Produto:
    
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    
    def etiqueta(self):
        return f'''
_____________Produto_____________
{self.nome}


R${self.preco:,.2f}

_________________________________
            '''
```

```python
banana = Produto("banana", 20)

print(banana.etiqueta())
```

# Crie a classe churrasco
quantidade de pessoas no churrasco, quantidadede carne a ser comprada, custo total do churrasco, preço por pessoa

```python
class Churrasco:
    
    def __init__(self, titulo, qtdpessoas, carnekg = 0.5, precokg = 23):
        self.titulo = titulo
        self.qtdpessoas = qtdpessoas
        self.carnekg = carnekg
        self.precokg = precokg

    
    def precoporPessoa(self):
        
        precototal = (self.carnekg * self.qtdpessoas * self.precokg )

        return f"""
Analisando o {self.titulo} com {self.qtdpessoas} convidados
Preço do kilo da carne R${self.precokg:.2f}
Quantidade recomendada de carne {self.qtdpessoas * self.carnekg}
Preço total R${precototal:.2f}
Preço total por pessoa R${precototal / self.qtdpessoas:.2f}
        """



represa = Churrasco("Churras na Represa", 15)
```

```python
represa.precoporPessoa()
```
