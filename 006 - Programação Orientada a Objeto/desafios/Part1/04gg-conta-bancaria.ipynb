```python
class contaBancaria:
    """
Crie uma conta bancária que permite fazer saques e depósitos
    """

    def __init__(self, id, nome, saldo = 0):
        self.__id = id
        self.__titular = nome
        self.__saldo = saldo
        print(f"Conta criada com sucesso!")

    def __str__(self):
        return f"A conta {self.__id} de {self.__titular} tem como saldo R${self.__saldo:,.2f}"


    def deposito(self, valor):
        self.__saldo += valor
        print(f"DEPÓSITO DE R${valor:,.2f} REALIZADO COM SUCESSO")
    
    def saque(self, valor):
        if valor <= self.__saldo:
            self.__saldo -= valor
            print(f"SAQUE DE R${valor:,.2f} REALIZADO COM SUCESSO")
        else:
            print(f"SALDO PARA SAQUE INSUFICIENTE")
```

# testes

```python
lysbradesco = contaBancaria(12345, "Lys", 50000)

print(lysbradesco) # ele imprime o __str__

print(lysbradesco.__doc__)
```

```python
lysbradesco.deposito(3000)

print(lysbradesco)

lysbradesco.saque(500)

print(lysbradesco)
```
