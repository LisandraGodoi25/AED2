```python
class Gafanhoto:
    #metodo construtor
    def __init__(self):
        self.nome = ""
        self.idade = 0

    #metodo de instancia
    def aniversario(self):
        self.idade += 1

    def mensagem(self):
        return f"{self.nome} é um gafanhoto e tem {self.idade} anos de idade"


    
```

# Melhorado

```python
class Gafanhoto2:
    """
    Essa classe cria um gafanhoto, para criar um novo gafanhoto, use:

    variavel = Gafanhoto2(nome, idade)
    """
    #metodo construtor
    def __init__(self, n = "", i = 0):
        self.__nome = n
        self.__idade = i

    #metodo de instancia
    def aniversario(self):
        self.__idade += 1

    def __getstate__(self):
        return f"{self.__nome} é um gafanhoto e tem {self.__idade} anos de idade"
    
```

```python
G1 = Gafanhoto2("Lys", 25)

G1.aniversario()
```
