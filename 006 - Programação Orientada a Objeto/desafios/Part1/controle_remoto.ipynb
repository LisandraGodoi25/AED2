```python
from rich import print
from rich.panel import Panel

class ControleRemoto:
    canal_min = 1
    canal_max = 7
    vol_min = 1
    vol_max = 9

    def __init__(self, canal = 1, vol = 5):
        self.canalatual = canal
        self.volumeatual = vol
        self.ligado:bool = False

    def ligaDesliga(self):
        self.ligado = not self.ligado

    def mostrarTV(self):
        if not self.ligado:
            conteudo = f":prohibited: [red]A TV está desligada[/] :prohibited:"
        else:
            conteudo = f"CANAL = "
            for canal in range(self.canal_min, self.canal_max+1):
                if canal == self.canalatual:
                    conteudo += f"[white on yellow] {canal} [/]"
                else:
                    conteudo += f" {canal} "

            conteudo += f"\n\nVOLUME = "
            for vol in range(self.vol_min, self.vol_max+1):
                if vol <= self.volumeatual:
                    conteudo += f"[cyan on cyan]{vol}[/]"
                else:
                    conteudo += f"[white on white]{vol}[/]"

        tv = Panel(conteudo, title = "TV", width=35)
        print(tv)
```

```python
c = ControleRemoto()

c.mostrarTV()

c.ligaDesliga()

c.mostrarTV()
```
