# Projeto 1: Biblioteca de Livros 
Contexto do mundo real: Imagine um sistema para gerenciar o empréstimo de livros em uma pequena biblioteca escolar. Queremos cadastrar livros e usuários, registrar quando alguém empresta ou devolve um 
livro. 

Descrição do problema: Cada Livro tem título, autor, quantidade disponível. Cada Usuário tem nome, 
matrícula e lista de livros emprestados. O sistema deve permitir que um usuário faça empréstimos (reduzindo a quantidade disponível) e devoluções (aumentando quantidade). Devemos evitar empréstimos se o livro estiver indisponível. Também precisamos manter os dados organizados e seguros (ex.: não permitir mexer diretamente na quantidade disponível).
