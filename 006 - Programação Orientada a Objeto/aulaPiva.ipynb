```python
class Stack:
    # funcao construtora
    def __init__(self):
        self.itens = []

    def is_empty(self):
        return self.itens == []

    def push(self, item):
        self.itens.append(self.item)
        print(f"Item {item} inserido!")

    def pop(self):
        item = self.itens.pop()
        return f"Item {item} removido!"

    def peek(self):
        return self.itens(-1)

    def size(self):
        return len(self.itens)

    def print(self):
        return print(self.itens)
```
