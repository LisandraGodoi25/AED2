```python
from rich import print
from abc import ABC, abstractclassmethod # abstract basic classes

class Pessoa(ABC):

    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def fazerAniversario(self):
        self.idade += 1

    @abstractclassmethod 
    def estudar(self):
        pass

# filho de pessoa
class Aluno(Pessoa):
    def __init__(self, nome, idade, curso, turma):
        super().__init__(nome, idade)
        self.curso = curso
        self.turma = turma

    def fazerMatricula(self):
        print(f"[black on white]{self.nome}[/] se matriculou com sucesso :green_circle:")

    def estudar(self):
        print("Estudando matemática")


# filho de pessoa
class Professor(Pessoa):
    def __init__(self, nome, idade, especialidade, nivel):
        super().__init__(nome, idade)
        self.especialidade = especialidade
        self.nivel = nivel

    def darAula(self):
        print(f"[yellow]{self.nome}[/] está dando aula :yellow_circle:") 

    def estudar(self):
        print("Estudando para dar aula de inglês")


# filho de pessoa
class Funcionario(Pessoa):
    def __init__(self, nome, idade, cargo, setor):
        super().__init__(nome, idade)
        self.cargo = cargo
        self.setor = setor

    def baterPonto(self):
        print(f"[yellow]{self.nome}[/] bateu ponto :blue_circle:")

    def printQueue(self):
        print(self.itens)

    def estudar(self):
        print("Estudando método de cozinhar mais rápido")
```
