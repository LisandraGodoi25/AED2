```python
from abc import ABC, abstractclassmethod

class Projeto(ABC):
    def __init__(self, cor): # ABSTRAÇÃO 
        self.__cor = cor
        self.status = "On Going"

    @abstractclassmethod
    def fecharProjeto(self):
        pass

    def fazendoProjeto(self):
        print("Fazendo projeto")

    # getter
    @property
    def showCor(self):
        print(self.__cor)
    
    # setter
    showCor.setter
    def changeColor(self, newcolor): # ENCAPSULAMENTO
        self.__cor = newcolor

class subProjeto(Projeto): # HERANÇA
    def __init__(self, cor):
        super().__init__(cor)

    def fecharProjeto(self):
        print("Fechando sub projeto") # POLIMORFISMO
```

```python
projetin = subProjeto("Azul")

projetin.fazendoProjeto()

projetin.fecharProjeto()
```

```python
projetin.showCor
```

```python
projetin.changeColor("Vermelho")

projetin.showCor
```
