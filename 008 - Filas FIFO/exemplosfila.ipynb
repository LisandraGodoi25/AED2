```python
from rich import print

class Queue:
    def __init__(self, limitItems):
        self.limitItems = limitItems
        self.Items = []
    
    def isEmpty(self):
        return True if self.Items == [] else False

    def enqueue(self, newitem):
        self.Items.insert(0, newitem)
        print(f"Elemento [yellow]{newitem}[/] foi inserido com sucesso! :green_circle:")

    def dequeue(self):
        return self.Items.pop()

    def size(self):
        return len(self.Items)
    
```

```python
teste = Queue(10)

teste.enqueue(2)
```
