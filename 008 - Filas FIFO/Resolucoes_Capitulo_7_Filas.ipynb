# Filas (Queues)

## Classe de Apoio: `Queue`
Antes de resolver os exercícios, precisamos definir a estrutura de **Fila (`Queue`)** utilizando `collections.deque`. O `deque` permite inserir no final (`append`) e remover do início (`popleft`) com complexidade de tempo constante $O(1)$, evitando o custo de deslocamento de elementos de uma lista comum.


```python
from collections import deque

class Queue:
    def __init__(self):
        # Inicializa uma fila vazia usando deque (double-ended queue)
        self.itens = deque()

    def is_empty(self):
        # Retorna True se a fila não tiver nenhum elemento, e False caso contrário
        return len(self.itens) == 0

    def enqueue(self, item):
        # Insere um novo elemento no FINAL da fila (complexidade O(1))
        self.itens.append(item)

    def dequeue(self):
        # Verifica se a fila está vazia antes de tentar remover
        if self.is_empty():
            return None
        # Remove e retorna o elemento que está na FRENTE (início) da fila (complexidade O(1))
        return self.itens.popleft()

    def peek(self):
        # Apenas consulta quem está na frente da fila sem removê-lo
        if self.is_empty():
            return None
        return self.itens[0]

    def size(self):
        # Retorna a quantidade atual de elementos na fila
        return len(self.itens)

print("Classe Queue carregada com sucesso!")
```

### Exercício 1 — Compreendendo o princípio FIFO (Qual elemento sai primeiro?)
**Nível:** ENADE | **Objetivo:** Reconhecer que `dequeue` remove o elemento mais antigo da fila.

**Enunciado:**
Em uma fila (estrutura de dados), é correto afirmar que:
* **A.** O último elemento que entra é o primeiro a sair.
* **B.** Não é possível inserir elementos enquanto realizamos remoções.
* **C.** O primeiro enfileiramento insere no início e o primeiro desenfileiramento remove do final.
* **D.** O elemento removido numa operação `dequeue` é sempre o mais antigo, isto é, o primeiro a ter sido enfileirado.
* **E.** Filas são acessadas apenas pelo meio da estrutura.


```python
# Demonstração prática do Exercício 1
fila_ex1 = Queue()

# Inserindo elementos na ordem: 1º, 2º e 3º
fila_ex1.enqueue("1º a chegar (Mais antigo)")
fila_ex1.enqueue("2º a chegar")
fila_ex1.enqueue("3º a chegar (Mais recente)")

# Ao chamar dequeue(), quem sai é o primeiro que entrou (regra FIFO)
removido = fila_ex1.dequeue()
print(f"Elemento removido pelo dequeue(): {removido}")

# Resposta correta da questão:
resposta_ex1 = "D"
print(f"Resposta correta: Alternativa {resposta_ex1} (First In, First Out - FIFO)")
```

### Exercício 2 — Acompanhando `enqueue`, `dequeue` e `peek`
**Nível:** ENADE | **Objetivo:** Determinar corretamente o elemento que permanece na frente após uma remoção.

**Enunciado:**
Fila inicialmente vazia: `enqueue(10)`, `enqueue(20)`, `dequeue()`, `enqueue(30)`, `peek()`. Qual valor é retornado por `peek()`, supondo que ele consulta a frente sem remover?
* **A.** 10
* **B.** 20
* **C.** 30
* **D.** Nenhum, pois a fila está vazia.
* **E.** Depende da implementação da fila.


```python
# Passo a passo das operações do Exercício 2
fila_ex2 = Queue()
print("Estado inicial:", list(fila_ex2.itens))

# 1. Insere 10 no final
fila_ex2.enqueue(10)
print("Após enqueue(10):", list(fila_ex2.itens))

# 2. Insere 20 no final (atrás do 10)
fila_ex2.enqueue(20)
print("Após enqueue(20):", list(fila_ex2.itens))

# 3. Remove o elemento da frente (o 10, que chegou primeiro)
removido = fila_ex2.dequeue()
print(f"Após dequeue() [removeu {removido}]:", list(fila_ex2.itens))

# 4. Insere 30 no final (atrás do 20)
fila_ex2.enqueue(30)
print("Após enqueue(30):", list(fila_ex2.itens))

# 5. Consulta quem está na frente sem remover
frente = fila_ex2.peek()
print(f"Valor retornado por peek(): {frente}")
print("Estado final da fila (peek não remove):", list(fila_ex2.itens))
print("Resposta correta: Alternativa B")
```

### Exercício 4 — Fila de Impressão em um Escritório
**Nível:** Prática | **Objetivo:** Simular processamento FIFO com uma pausa obrigatória após cada três impressões.

**Enunciado:**
Implemente uma fila de impressão. Cada documento possui um nome e deve ser processado na ordem de chegada. Após cada três documentos impressos, a impressora precisa ficar um ciclo sem processar novos documentos para resfriamento; durante esse ciclo, novos documentos ainda podem entrar na fila.


```python
def fila_impressao(documentos, chegadas_por_ciclo=None):
    # Garante que chegadas_por_ciclo seja um dicionário (evita erro se vier None)
    chegadas_por_ciclo = chegadas_por_ciclo or {}
    fila = Queue()

    # 1. Enfileira todos os documentos iniciais na ordem de chegada
    for documento in documentos:
        fila.enqueue(documento)

    impressos_desde_pausa = 0  # Contador de documentos impressos no bloco atual
    resfriando = False         # Flag booleana para controlar o ciclo único de pausa
    ciclo = 1                  # Contador de ciclos de execução

    # O laço continua enquanto houver documentos na fila, estiver resfriando
    # ou ainda houver documentos programados para chegar em ciclos futuros
    while not fila.is_empty() or resfriando or ciclo in chegadas_por_ciclo:
        print(f"--- Ciclo {ciclo} ---")
        
        # 2. Primeiro verificamos se chegam novos documentos neste ciclo.
        # Isso acontece INDEPENDENTE de a impressora estar resfriando ou não.
        for documento in chegadas_por_ciclo.get(ciclo, []):
            fila.enqueue(documento)
            print(f"Novo documento recebido: {documento}")

        # 3. Verifica o estado da impressora
        if resfriando:
            # Se estava marcada para resfriar, gasta este ciclo apenas descansando
            print("Pausa para resfriamento da impressora.")
            resfriando = False  # Libera a impressora para o próximo ciclo
        elif not fila.is_empty():
            # Se não está resfriando e há documentos, imprime o mais antigo (FIFO)
            atual = fila.dequeue()
            print(f"Imprimindo documento: {atual}")
            impressos_desde_pausa += 1

            # Se atingiu 3 impressões seguidas, ativa o resfriamento para o próximo ciclo
            if impressos_desde_pausa == 3:
                resfriando = True
                impressos_desde_pausa = 0  # Zera o contador para o próximo lote

        ciclo += 1

# Testando a função com 6 documentos iniciais e 1 chegando no ciclo 4 (durante a pausa!)
fila_impressao(
    ["Relatório.pdf", "Contrato.docx", "Imagem.png", "Resumo.txt", "Tabela.xlsx", "Mapa.jpg"],
    chegadas_por_ciclo={4: ["Ata.docx"]}
)
```

### Exercício 5 — Gerenciador de Senhas Preferenciais e Comuns
**Nível:** Prática | **Objetivo:** Coordenar duas filas FIFO com uma regra simples de prioridade de atendimento.

**Enunciado:**
Use duas filas: uma para senhas preferenciais e outra para senhas comuns. Em cada ciclo, atenda primeiro uma senha preferencial, se houver, e depois uma senha comum, se houver. Repita até ambas ficarem vazias.


```python
def gerenciador_senhas(preferenciais, comuns):
    # Criamos duas filas independentes para preservar a ordem de chegada (FIFO) dentro de cada grupo
    fila_pref = Queue()
    fila_comum = Queue()

    # Preenche a fila preferencial
    for senha in preferenciais:
        fila_pref.enqueue(senha)

    # Preenche a fila comum
    for senha in comuns:
        fila_comum.enqueue(senha)

    ciclo = 1
    # Repete enquanto pelo menos uma das filas ainda tiver pessoas aguardando
    while not fila_pref.is_empty() or not fila_comum.is_empty():
        print(f"--- Rodada de Atendimento {ciclo} ---")
        
        # 1º passo do ciclo: atende uma senha preferencial, se existir
        if not fila_pref.is_empty():
            senha = fila_pref.dequeue()
            print(f"Atendendo (preferencial): {senha}")

        # 2º passo do ciclo: atende uma senha comum, se existir
        if not fila_comum.is_empty():
            senha = fila_comum.dequeue()
            print(f"Atendendo (comum): {senha}")
            
        ciclo += 1

# Exemplo de execução
gerenciador_senhas(
    ["P01", "P02", "P03"],
    ["C01", "C02", "C03", "C04"]
)
```

### Projeto 1 — Gestão de Tarefas
**Objetivo:** Simular uma fila corporativa de pendências, processando uma tarefa por ciclo e aceitando novas entradas.

**Problema:**
Iniciar com tarefas pendentes, concluir uma tarefa por vez, informar qual foi processada e permitir que novas tarefas sejam acrescentadas depois de cada ciclo.


```python
def gestao_tarefas(tarefas_iniciais, ciclos, novas_por_ciclo=None):
    novas_por_ciclo = novas_por_ciclo or {}
    fila = Queue()

    # Carrega as tarefas iniciais na ordem em que foram registradas
    for tarefa in tarefas_iniciais:
        fila.enqueue(tarefa)

    # Executa a simulação pelo número definido de ciclos
    for ciclo in range(1, ciclos + 1):
        print(f"--- Ciclo {ciclo} ---")
        
        # 1. Processa (conclui) a tarefa mais antiga da fila
        if fila.is_empty():
            print("Não há tarefa pendente para processar.")
        else:
            atual = fila.dequeue()
            print(f"Tarefa concluída: {atual}")

        # 2. Após o processamento do ciclo, novas tarefas entram no FINAL da fila
        # Elas deverão aguardar todas as tarefas anteriores serem concluídas
        for nova in novas_por_ciclo.get(ciclo, []):
            fila.enqueue(nova)
            print(f"Tarefa adicionada: {nova}")

# Exemplo de execução: 3 tarefas iniciais, 5 ciclos e chegadas programadas nos ciclos 1 e 3
gestao_tarefas(
    tarefas_iniciais=["Relatório", "E-mail", "Backup"],
    ciclos=5,
    novas_por_ciclo={
        1: ["Atualizar dashboard"],
        3: ["Conferir estoque"]
    }
)
```

### Projeto 2 — Fila de Cinema
**Objetivo:** Simular a compra de ingressos em sessões sucessivas, com atendimentos FIFO e chegada de novos clientes.

**Problema:**
A partir de uma fila inicial, realizar um número definido de atendimentos, imprimir cada cliente atendido e acrescentar novas pessoas ao fim da fila no início de determinadas sessões.


```python
def fila_cinema(clientes_iniciais, total_atendimentos, atendimentos_por_sessao, chegadas_por_sessao=None):
    chegadas_por_sessao = chegadas_por_sessao or {}
    fila = Queue()

    # Enfileira os clientes que já estavam na bilheteria antes da Sessão 1
    for cliente in clientes_iniciais:
        fila.enqueue(cliente)

    atendidos = 0  # Controla o limite máximo de atendimentos da simulação
    sessao = 1     # Controla o número da sessão atual

    # Continua abrindo sessões até atingir o total de atendimentos planejado
    while atendidos < total_atendimentos:
        print(f"--- Sessão {sessao} ---")
        
        # 1. No início da sessão, novos clientes chegam e vão para o FINAL da fila
        for novo in chegadas_por_sessao.get(sessao, []):
            fila.enqueue(novo)
            print(f"Chegou à fila: {novo}")

        # Se não houver ninguém na fila mesmo após as chegadas, encerra a simulação
        if fila.is_empty():
            print("Não há clientes aguardando.")
            break

        # 2. Realiza os atendimentos desta sessão respeitando dois limites:
        #    - o limite de atendimentos por sessão
        #    - o limite total de atendimentos da simulação
        for _ in range(atendimentos_por_sessao):
            if fila.is_empty() or atendidos == total_atendimentos:
                break
            
            # Retira o cliente da frente da fila (quem esperou mais tempo)
            cliente = fila.dequeue()
            print(f"{cliente} comprou ingresso e entrou.")
            atendidos += 1

        sessao += 1

# Exemplo de execução
fila_cinema(
    clientes_iniciais=["Cliente A", "Cliente B", "Cliente C"],
    total_atendimentos=5,
    atendimentos_por_sessao=2,
    chegadas_por_sessao={
        2: ["Cliente D", "Cliente E"],
        3: ["Cliente F"]
    }
)
```
