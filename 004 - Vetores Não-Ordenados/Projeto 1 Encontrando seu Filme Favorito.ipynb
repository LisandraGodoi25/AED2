```python
def findMovie():
    aux = 0
    tofind = input("Escreva o termo de pesquisa: ").capitalize()

    print("TERMO PROCURADO: ")
    print(tofind)

    filteredlist = []

    for indice in range(len(filmes)):
        #print(filmes[indice])
        if filmes[indice].find(tofind) >= 0:
            #print("ENCONTRADO")
            filteredlist.append(filmes[indice])
            #print(filteredlist)
        aux += 1

    if len(filteredlist)>0:
        print("FILMES ENCONTRADOS: ")

        for movie in filteredlist:
            print(movie)
    else:
        print("NENHUM FILME ENCONTRADO")
```
