4. (PRÁTICA): Implemente uma função em Python chamada buscar_cliente(nome, lista_clientes) que 
realiza busca sequencial em uma lista de nomes de clientes. Essa função deve retornar o índice onde o nome foi encontrado ou -1 caso o nome não esteja na lista_clientes. Em seguida, escreva um pequeno código de exemplo que utilize essa função para buscar por pelo menos um nome que exista e um que não exista na lista, exibindo mensagens adequadas. 


```python
nomes = ["Lisandra", "Julia","Lorena","Kamilly","Paulo","Gustavo","Giovanni"]
```

```python
def buscar_cliente(nome, lista_clientes):
    aux = 0
    for name in lista_clientes:
        if name == nome:
            print(aux)
            break
        else:
            aux += 1
            continue
        
    if nome not in lista_clientes:
        print("-1")

```

```python
buscar_cliente("Ana",nomes)
```

```python
buscar_cliente("Paulo",nomes)
```

5. (PRÁTICA):  Crie uma função conta_ocorrencias(valor, lista) em Python que percorre uma lista e conta 
quantas  vezes  um  determinado  valor  aparece.  Utilize  busca  sequencial  (varrendo  a  lista)  para  realizar  a contagem. Por exemplo, conta_ocorrencias(3, [1,3,5,3,7]) deve retornar 2, pois o valor 3 aparece duas vezes na lista. Inclua casos de teste chamando a função com diferentes valores e listas. 


```python
numlist = []

def criar_lista():
    while True:
        addNum = int(input("Insira um número inteiro ou -1 para parar: "))
        if addNum == -1:
            break
        else:
            numlist.append(addNum)
    
    print(numlist)

criar_lista()
```

```python
def conta_ocorrencias(valor, lista):
    aux = 0
    for nums in lista:
        if valor == nums:
            aux += 1

    return aux 
```

```python
conta_ocorrencias(1,numlist)
```

6. (PRÁTICA): Suponha que precisamos remover um elemento específico de uma lista não ordenada. Escreva 
uma função Python remove_elemento(lista, valor) que remove a primeira ocorrência de valor na lista, 
mantendo os demais elementos e retornando o valor removido. A função deve retornar None caso o valor não 
seja encontrado ou a lista esteja vazia. Dica: Utilize a busca sequencial para encontrar a posição do elemento e,  se  encontrada,  deslize  os  elementos  posteriores  para  “preencher”  o  espaço  removido  (ou  use  métodos próprios da lista Python). Demonstre o funcionamento da função com um exemplo. 

```python
numlistremove = []

def criar_lista():
    while True:
        addNum = int(input("Insira um número inteiro ou -1 para parar: "))
        if addNum == -1:
            break
        else:
            numlistremove.append(addNum)
    
    print(numlistremove)

criar_lista()
```

```python
def remove_elemento(lista, valor):
    for nums in range(len(lista)):
        if lista[nums] == valor:
            print(f"Remover item: {lista[nums]}")
            lista.pop(nums)
            break

    print(f"Lista atual: {lista}")
```

```python
remove_elemento(numlistremove, 2)
```
