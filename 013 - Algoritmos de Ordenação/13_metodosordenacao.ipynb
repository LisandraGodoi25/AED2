## Exercício 7 - Passo a passo do Bubble Sort (Prática)

**Enunciado:**
Considere $X = [7, 4, 5, 2]$. Trace o passo a passo de como o Bubble Sort ordenaria `X` em ordem crescente, mostrando o estado do vetor após cada passagem completa.

```python
def bubble_sort_passos(arr):
    lista = arr.copy()
    n = len(lista)
    print(f"Vetor inicial: {lista}")
    for i in range(n - 1):
        trocou = False
        for j in range(n - 1 - i):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                trocou = True
        print(f"Após a {i + 1}ª passagem: {lista}")
        if not trocou:
            break
    return lista

X = [7, 4, 5, 2]
resultado = bubble_sort_passos(X)
print(f"Vetor final: {resultado}")
```

## Projeto 1 - Análise de Vendas Ordenadas

**Enunciado:**
Desenvolver um código que ordene a lista de faturamentos mensais em ordem decrescente utilizando Quicksort e identifique o maior e o menor valor.

```python
# Faturamentos mensais, em milhares de reais
vendas = [120, 85, 90, 100, 75, 130, 95, 110, 80, 140, 70, 105]

def particao_decrescente(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] >= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort_decrescente(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low < high:
        pi = particao_decrescente(arr, low, high)
        quick_sort_decrescente(arr, low, pi - 1)
        quick_sort_decrescente(arr, pi + 1, high)
    return arr

vendas_ordenadas = quick_sort_decrescente(vendas.copy())

print("Vendas mensais (original):", vendas)
print("Vendas ordenadas (decrescente):", vendas_ordenadas)
print("Melhor mês (maior venda):", vendas_ordenadas[0])
print("Pior mês (menor venda):", vendas_ordenadas[-1])
```

## Projeto 2 - Jogo de Ordenação de Cartas

**Enunciado:**
Gerar uma pequena sequência de valores, tratar a lista de compras como pilha e inserir cada carta na posição correta de uma lista ordenada (simulando a lógica incremental do Insertion Sort).

*(Nota: o `input()` interativo foi deixado comentado abaixo para permitir execução direta no notebook; descomente se quiser pausar a cada rodada).*

```python
import random

random.seed(42)  # Semente opcional para reprodutibilidade
cartas = [random.randint(1, 99) for _ in range(5)]
pilha_compras = cartas.copy()
random.shuffle(pilha_compras)

mao = []
print("Cartas embaralhadas na pilha:", pilha_compras)

while pilha_compras:
    # input("\nPressione Enter para comprar uma carta...")
    carta = pilha_compras.pop()
    print(f"\nVocê comprou a carta: {carta}")

    if not mao:
        mao.append(carta)
    else:
        pos = 0
        while pos < len(mao) and mao[pos] < carta:
            pos += 1
        mao.insert(pos, carta)

    print("Sua mão agora:", mao)

print("\nTodas as cartas foram compradas e ordenadas na mão!")
```
