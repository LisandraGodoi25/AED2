# Projeto 2: A Lista de Tarefas: 

• Contexto: Você precisa criar uma lista de tarefas para a semana. 

• Problema: Como você pode adicionar novas tarefas, marcar tarefas como concluídas e visualizar as tarefas pendentes? 

• Indicação de como resolver: Pensar em uma forma de manter uma sequência de itens que podem ser adicionados e removidos. 

## Creating list

```python
taskList = ["Ir ao supermercado", "Terminar a lição de casa"]
```

## Funções

```python
def showTasks():
    print("TASK LIST")
    aux = 1
    for posicao in taskList:
        print(f"{aux} - {posicao}")
        aux += 1
```

```python
showTasks()
```

```python
def addTask():
    while True:
        inputTask = input("Add new task or Q to stop: ").capitalize()
        if inputTask == "Q":
            break
        else:
            taskList.append(inputTask)

    showTasks()
```

```python
addTask()
```

```python
def markCompleted():

    todelete = int(input("Choose the number of the task you want to mark as complete or Q to stop: "))
    taskList.pop(todelete-1)
    print("Task completed. Good job!")
```

```python
markCompleted()
showTasks()
```

# Main

```python
while True:

    print('''

TASK LIST

    [A]Add new task [B] Mark a task as completed [C] View all open task [Q] Quit
    ''')

    selectedoption = input("Go to: ").upper()


    if selectedoption == "A":
        addTask()

    elif selectedoption == "B":
        markCompleted()

    elif selectedoption == "C":
        showTasks()

    elif selectedoption == "Q":
        print("Bye, bye!")
        break
    else:
        continue
```
