Exercício 1: Crie uma lista em Python chamada frutas com os seguintes elementos: "maçã", "banana", "laranja". Em seguida, adicione a fruta "morango" ao final da lista. Por fim, imprima a lista completa. 

```python
frutas = ["maçã", "banana", "laranja"]

#print(frutas)

frutas.append("morango")
aux = 1

for item in frutas:

   print(f"{aux}. {item}")
   aux += 1
```

Exercício 3: Crie uma fila (usando uma lista Python) chamada clientes. Adicione os clientes "Ana", "Bruno" e "Carla" nesta ordem. Em seguida, simule a chegada de um novo cliente "Daniel" e a saída do primeiro cliente da fila. Imprima o cliente que saiu e a fila restante. 

```python
clientes = ["Ana", "Bruno" , "Carla"]

primeiro = clientes[0]

#adicione o novo cliente Daniel
clientes.append("Daniel")

#Saida do primeiro cliente
clientes.pop(0)

print(f"O que cliente que saiu foi {primeiro}")

print("Os clientes restantes são:")
    
for pessoa in clientes:
    print(f"{pessoa}")
```
