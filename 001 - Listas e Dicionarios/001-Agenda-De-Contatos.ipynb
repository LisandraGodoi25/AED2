# Projeto 1: A Agenda de Contatos: 

• Contexto: Você precisa criar um sistema simples para armazenar e gerenciar contatos (nome, telefone e e-mail). 

• Problema: Como você pode adicionar novos contatos, buscar um contato pelo nome e listar todos os contatos? 

• Indicação de como resolver: Pensar em uma estrutura que permita associar informações (telefone e e-mail) a um nome de forma eficiente.

## Estrutura do dicionário

```python
#Estrutura do dicionário

contatos = {"Lisandra" : 
                {
                "Telefone": "1345-6432", 
                "Email" : "lisandrafernandadegodoi@gmail.com"}}
```

## Funções

```python
def addContacts(addName, addTelefone, addEmail):
    contatos[addName] = {"Telefone" : addTelefone, "Email":addEmail}

    print(f"O contato de {addName} foi adicionado com sucesso!")
```

```python
def showContacts():
    for posicao in range(len(contatos)):
        print( f"{list(contatos.keys())[posicao]}: ")
        print(f"Telefone: {contatos[list(contatos.keys())[posicao]]['Telefone']}")
        print(f"Email: {contatos[list(contatos.keys())[posicao]]['Email']}")
        print("")

        
```

```python
def filterContacts():
     
     searchterm = input("Insira o nome do contato: ")

     for posicao in range(len(contatos)):
        if searchterm in list(contatos.keys())[posicao]:
            print("CONTATOS ENCONTRADOS: ")
            print( f"{list(contatos.keys())[posicao]}: ")
            print(f"Telefone: {contatos[list(contatos.keys())[posicao]]['Telefone']}")
            print(f"Email: {contatos[list(contatos.keys())[posicao]]['Email']}")
            print("")
```

## Populando o dicionário

```python
#Adicionar contatos para teste

addContacts("Julia","12346543","juliaamaue@gmail.com")
addContacts("Kamilly","4323256","kamillyyy@gmail.com")
addContacts("Paulo","859449373","paulocesar@gmail.com")
addContacts("Gustavo","065834723","reisgust@gmail.com")
addContacts("Lorena","2195934838","lorenacristiane@gmail.com")
```

# Main

```python
while True:

    print('''

CONTACT LIST

    [A] View Contact List [B] Search for a contact [C] Add a new contact [Q] Quit
    ''')

    selectedoption = input("Go to: ").upper()


    if selectedoption == "A":
        showContacts()

    elif selectedoption == "B":
        filterContacts()

    elif selectedoption == "C":
        inputName = input("Name: ")
        inputTelefone = input("Telefone: ")
        inputEmail = input("Email: ")
        addContacts(inputName, inputTelefone, inputEmail)

    elif selectedoption == "Q":
        print("Adios!")
        break
    else:
        continue
```
