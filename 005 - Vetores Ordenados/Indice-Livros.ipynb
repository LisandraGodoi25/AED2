```python
def FindBooks(listbooks):
    tofind = input("Coloque o nome do item qu deseja procurar: ").upper()

    aux = 0
    first = 0
    
    last = len(listbooks)-1
    #print(f"{last}")



    while first <= last:
        aux += 1
        #print(f"first: {first}")
        #print(f"last: {last}")
        meio = (first + last)// 2 
        #print(f"meio: {meio}")
        #print(f"Livro do meio é {listbooks[meio]}")

        #print(f"O meu {listbooks[meio]}")
        if listbooks[meio][0].upper() == tofind:
            print(F"{tofind}")
            print(F"Autor: {listbooks[meio][1]}")
            print(F"Ano de publicação: {listbooks[meio][2]}")

        elif tofind < listbooks[meio][0].upper():

            last = meio - 1

        else:
            
            first = meio + 1


    #print(f"Quantidade de passos: {aux}")
    return "Item não encontrado"




```
