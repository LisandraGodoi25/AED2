Contexto: Imagine que você está desenvolvendo a funcionalidade de busca para um site de e-commerce com um grande número de produtos. O catálogo de produtos é armazenado em uma lista, onde cada item é um dicionário contendo informações como nome, preço e descrição. # Função

```python
def findItem(listItems):
    tofind = input("Coloque o nome que deseja procurar").upper()

    aux = 0
    firstItem = 0
    lastItem = len(listItems)-1




    while firstItem <= lastItem:
        aux += 1
        meio = (firstItem + lastItem)// 2 

        if listItems[meio]['Nome'].upper() == tofind:
            print(F"{tofind}")
            print(F"R${listItems[meio]['Preço']}")
            print(F"{listItems[meio]['Descrição']}")
            return meio

        elif tofind < listItems[meio]['Nome'].upper():

            lastItem = meio - 1

        else:
            
            firstItem = meio + 1


    print(f"Quantidade de passos: {aux}")
    return "Item não encontrado"



```
