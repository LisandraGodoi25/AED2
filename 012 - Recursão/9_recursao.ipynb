## Exercício 3 - Soma dos dígitos (Prática)

**Enunciado:**
Escreva uma função recursiva em Python chamada `soma_digitos(n)` que calcule a soma dos dígitos de um número inteiro positivo `n`. Por exemplo, `soma_digitos(123)` deve retornar `6`.

````python
def soma_digitos(n):
    if n < 10:  # caso base: apenas um dígito
        return n
    ultimo_digito = n % 10
    resto_do_numero = n // 10
    return ultimo_digito + soma_digitos(resto_do_numero)

print("soma_digitos(123):", soma_digitos(123))
print("soma_digitos(5072):", soma_digitos(5072))
````

## Exercício 4 - Verificação de palíndromo (Prática)

**Enunciado:**
Escreva uma função recursiva em Python chamada `palindromo(s)` que verifique se uma string é um palíndromo. Ignore espaços e diferenças entre maiúsculas e minúsculas.

````python
def palindromo(s):
    s = s.lower().replace(" ", "")
    if len(s) <= 1:  # caso base
        return True
    if s[0] != s[-1]:
        return False
    return palindromo(s[1:-1])

print(palindromo("A man a plan a canal Panama"))
print(palindromo("Socorram me subi no onibus em Marrocos"))
print(palindromo("hello"))
````

## Exercício 5 - Convertendo uma soma iterativa em recursiva (Prática)

**Enunciado:**
Converta a função iterativa abaixo, que calcula a soma dos números de `1` a `n`, em uma função recursiva:
```python
def soma_iterativa(n):
    soma = 0
    for i in range(1, n + 1):
        soma += i
    return soma
```

````python
def soma_recursiva(n):
    if n == 1:  # caso base
        return 1
    return n + soma_recursiva(n - 1)

print("soma_recursiva(5):", soma_recursiva(5))
````

## Projeto 1 - Analisando uma Estrutura Organizacional Hierárquica

**Enunciado:**
Escrever uma função recursiva que receba o dicionário da estrutura organizacional e retorne o total de funcionários da organização inteira, incluindo todos os níveis hierárquicos.

````python
def contar_funcionarios(departamento):
    total = departamento.get("funcionarios", 0)
    subdepartamentos = departamento.get("subdepartamentos", [])
    for subdepartamento in subdepartamentos:
        total += contar_funcionarios(subdepartamento)
    return total

estrutura = {
    "nome": "Diretoria",
    "funcionarios": 10,
    "subdepartamentos": [
        {
            "nome": "Financeiro",
            "funcionarios": 5
        },
        {
            "nome": "Tecnologia",
            "funcionarios": 8,
            "subdepartamentos": [
                {"nome": "Dados", "funcionarios": 3},
                {"nome": "Infraestrutura", "funcionarios": 2}
            ]
        }
    ]
}

print(f"Total de funcionários: {contar_funcionarios(estrutura)}")
````

## Projeto 2 - Buscando Arquivos por Extensão em um Sistema de Arquivos

**Enunciado:**
Escrever uma função recursiva que receba um caminho de diretório e uma extensão e retorne uma lista com os caminhos completos de todos os arquivos encontrados com essa extensão.

````python
import os
import tempfile

def encontrar_arquivos(diretorio, extensao):
    arquivos_encontrados = []
    try:
        for item in os.listdir(diretorio):
            caminho_completo = os.path.join(diretorio, item)
            if os.path.isfile(caminho_completo):
                if caminho_completo.lower().endswith(extensao.lower()):
                    arquivos_encontrados.append(caminho_completo)
            elif os.path.isdir(caminho_completo):
                encontrados_no_subdiretorio = encontrar_arquivos(
                    caminho_completo, extensao
                )
                arquivos_encontrados.extend(encontrados_no_subdiretorio)
    except FileNotFoundError:
        print(f"Erro: diretório '{diretorio}' não encontrado.")
    return arquivos_encontrados

# Criando estrutura de exemplo temporária para demonstração executável
with tempfile.TemporaryDirectory() as tmpdir:
    dados_dir = os.path.join(tmpdir, "dados")
    sub_dir = os.path.join(dados_dir, "2026")
    os.makedirs(sub_dir)
    for f in ["vendas.csv", "clientes.xlsx"]:
        open(os.path.join(dados_dir, f), "w").close()
    for f in ["janeiro.csv", "fevereiro.csv"]:
        open(os.path.join(sub_dir, f), "w").close()

    arquivos = encontrar_arquivos(dados_dir, ".csv")
    print("Arquivos .csv encontrados:")
    for arquivo in sorted(arquivos):
        print(os.path.relpath(arquivo, tmpdir))
````
