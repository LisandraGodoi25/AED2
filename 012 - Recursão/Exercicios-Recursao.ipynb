3. (PRÁTICA) Escreva uma função recursiva em Python chamada soma_digitos(n) que calcule a soma dos dígitos de um número inteiro positivo n. Por exemplo, soma_digitos(123) deve retornar 1 + 2 + 3 = 6

```python
def soma_digitos(numlong, soma = 0):
    unidade = numlong % 10
    resto = numlong // 10

    if unidade == 0 and resto == 0:
        return print(soma)

    soma += unidade

    soma_digitos(resto, soma)
```

```python
soma_digitos(123456789)
```

```python
soma_digitos(31784049074)
```
